```python
import pandas as pd
import numpy as np

data = {
    "product_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones",
             "Webcam", "Printer", "USB Cable", "Speaker", "Tablet"],
    "price": [55000, 800, 1500, 12000, 2500, 1800, 9000, 400, 3500, 25000],
    "qty": [2, 10, 5, 3, 8, 4, 2, 15, 6, 3]
}

df = pd.DataFrame(data)

df["Total_Price"] = df["price"] * df["qty"]

df["Discount"] = np.where(
    df["Total_Price"] > 5000,
    df["Total_Price"] * 0.10,
    df["Total_Price"] * 0.05
)

df["Free_Delivery"] = np.where(
    df["Total_Price"] > 2000,
    "Yes", "No"
)

df["GST"] = df["Total_Price"] * 0.08

df["Total_Bill"] = df["Total_Price"] - df["Discount"] + df["GST"]

print(df)
```
