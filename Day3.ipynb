```python
# Implementation of dictionary minor project

employee = dict()

while True:
    print("1. Add")
    print("2. View")
    print("3. Search")
    print("4. Update")
    print("5. Delete")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        id = int(input("Enter Employee ID: "))
        name = input("Enter Employee Name: ")
        age = input("Enter age: ")
        dept = input("Enter Department: ")

        employee[id] = {
            "name": name,
            "age": age,
            "dept": dept
        }

        print("Employee Added successfully")

    elif choice == 2:
        if employee:
            for id, data in employee.items():
                print("Emp ID:", id)
                print("Emp Name:", data["name"])
                print("Emp Age:", data["age"])
                print("Emp Department:", data["dept"])

        else:
            print("No record Found")

    elif choice == 3:
        eid = int(input("Enter your id: "))

        if eid in employee:
            print("Emp Name:", employee[eid]["name"])
            print("Emp Age:", employee[eid]["age"])
            print("Emp Dept:", employee[eid]["dept"])
        else:
            print("Employee Not found")

    elif choice == 4:
        eid = int(input("Enter id to update: "))

        if eid in employee:
            name = input("Enter your name: ")
            age = int(input("Enter your age: "))
            dept = input("Enter dept: ")

            employee[eid] = {
                "name": name,
                "age": age,
                "dept": dept
            }

            print("Employee updated successfully")

        else:
            print("Employee not found")

    elif choice == 5:
        eid = int(input("Enter id to remove: "))

        if eid in employee:
            employee.pop(eid)
            print("Employee deleted successfully")

        else:
            print("Employee not found")

    elif choice == 6:
        break
```

```python
# Dishes Minor Project

Dishes = {}

while True:
    print("\n1. Add Dish")
    print("2. View Menu")
    print("3. Search Dish")
    print("4. Update Price")
    print("5. Remove Dish")
    print("6. Exit")

    ch = int(input("Enter your Choice : "))

    
    if ch == 1:
        id = int(input("Enter Dish Id : "))
        name = input("Enter Dish Name : ")
        price = int(input("Enter Dish Price : "))

        Dishes[id] = {
            "name": name,
            "price": price
        }

        print("Dish Added Successfully")

    
    elif ch == 2:
        if Dishes:
            print("\n--- Menu ---")
            for id, dish in Dishes.items():
                print(id, " ", dish["name"], " ", dish["price"])
        else:
            print("No Dishes Added yet")

    
    elif ch == 3:
        did = int(input("Enter Dish Id : "))

        if did in Dishes:
            print("Dish Name :", Dishes[did]["name"])
            print("Dish Price :", Dishes[did]["price"])
        else:
            print("No Dishes Found")

    
    elif ch == 4:
        did = int(input("Enter Dish Id : "))

        if did in Dishes:
            price = int(input("Enter New Price : "))
            Dishes[did]["price"] = price
            print("Price Updated Successfully")
        else:
            print("No Dishes Found")

    
    elif ch == 5:
        did = int(input("Enter Dish Id : "))

        if did in Dishes:
            del Dishes[did]
            print("Dish Removed Successfully")
        else:
            print("No Dishes Found")


    elif ch == 6:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
