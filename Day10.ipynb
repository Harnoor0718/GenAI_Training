```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression # here linear Regression is a class
```

```python
data = {
    "study_hour" : [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "marks" : [23, 45, 57, 64, 69, 79, 88, 89, 97]
}

df = pd.DataFrame(data)
df
```

```python
plt.plot(df["study_hour"], df["marks"], color = "c", linestyle = "--")
sns.scatterplot(data=df, x="study_hour", y="marks", color = "y")
plt.xlabel("Study Hours ---->")
plt.ylabel("Marks ---->")
plt.plot()
```

```python
# data is divided into 2 parts for training experience
X = df[["study_hour"]]
y = df["marks"]
```

```python
# model build
model = LinearRegression()
model
```

```python
# model train
model.fit(X, y) # fit() is used for training experience
```

```python
c = model.intercept_
print(round(c, 2))
```

```python
# y = mx + c

m = model.coef_[0]
print(round(m, 2))
```

```python
# formula apply y = mx + c
study_hour = float(input("Enter study hours :"))
y = m * study_hour + c
print("Predicted Marks :", round(y, 2))
```
