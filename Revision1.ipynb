```python
class PoliceStation:

    class FIR:
        def __init__(self, fir_no, complaint, crime, location):
            self.fir_no = fir_no
            self.complaint = complaint
            self.crime = crime
            self.location = location


        def details(self):
            print("FIR No :", self.fir_no)
            print("Complaint :", self.complaint)
            print("Crime Type :", self.crime)
            print("Location :", self.location)


    def __init__(self):
        self.firs = []

    def register_fir(self):
        fir_no = int(input("Enter FIR No :"))
        complaint = input("Enter your Complaint :")
        crime = input("Enter the Type of Crime :")
        location = input("Enter crime Location :")

        f = self.FIR(fir_no, complaint, crime, location)
        self.firs.append(f)

        print("Fir Added SuccessFully")

    def show_firs(self):
        if len(self.firs) == 0:
            print("No FIR added yet")

        else:
            for i in self.firs:
                i.details()

    def search_fir(self):
        fir_no = int(input("Enter fir_no to search :"))

        found = False

        for i in self.firs:
            if i.fir_no == fir_no:
                i.details()
                found = True

        if found == False:
            print("No FIR Found")


    def update_fir(self):
        fir_no = int(input("Enter FIR No to update :"))

        found = False
        for i in self.firs:
            if i.fir_no == fir_no:
                found = True
                i.location = input("Enter new Location :")

                print("Location Updated Sucessfully")

        if found == False:
            print("No FIR Found")


    def del_fir(self):
        fir_no = int(input("Enter FIR No to Delete :"))

        found = False

        for i in self.firs:
            if i.fir_no == fir_no:
                found = True
                self.firs.remove(i)

                print("FIR deleted successfully")
                break

        if found == False:
            print("No FIR Found")


station = PoliceStation()

while True:

    print("\n===== POLICE STATION =====")
    print("1. Register FIR")
    print("2. Show All FIRs")
    print("3. Search FIR")
    print("4. Update FIR")
    print("5. Delete FIR")
    print("6. Exit")

    choice = int(input("Enter your choice : "))

    if choice == 1:
        station.register_fir()

    elif choice == 2:
        station.show_firs()

    elif choice == 3:
        station.search_fir()

    elif choice == 4:
        station.update_fir()

    elif choice == 5:
        station.del_fir()

    elif choice == 6:
        print("Thank you for using FIR Management System")
        break

    else:
        print("Invalid Choice")  


    


        

```
