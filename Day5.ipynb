```python
class Test:
    # class level variable
    a = 10
    def __init__(self):
        # instance variable
        self.b = 20
t = Test()
print(t.a, t.b)

t.a = 11
t.b = 22
print(t.a, t.b)

t1 = Test()
print(t1.a, t1.b) # instance variable is varied from object to object
```

```python
class Demo:

    def msg(self):
        print("bye")

    @classmethod
    def show(cls):
        print("hello")

    @staticmethod
    def disp():
        print("hi")

d = Demo()

# We can access instance method by using object reference
d.msg()

# We can access classmethod by using object reference as well as class name
Demo.show()
d.show()

# we can access static method by using only class name
Demo.disp()



    
```

```python
# when used instance, classmethod, staticmethod
# self can get object information
# cls can get class information
# staticmethod can get only logic and calculation

# we can declare static/ class level variable inside class and constructor

class Test:
    count = 0

    def __init__(self, name):
        self.name = name
        Test.count += 1

t1 = Test("Ravi")
t2 = Test("Mukesh")
t3 = Test("Rahul")
print("Number of employees", Test.count)
```

```python
# diff bw instance vs static/ class vs local variable

class Student:

    # class level

    clg_name = "CGC"

    def __init__(self, name, rollNo):

        # instance level
        
        self.name = name
        self.rollNo = rollNo

    def show(self):
        greeting = "hello, " + self.name
        print(greeting)
        print(self.clg_name, self.name, self.rollNo)

    

s = Student("Harnoor", 89)
s.show()







        
```

```python
class A:

    def m1(self):
        print("hello")

    def m2(self, a, b):
        print(a + b)


class B(A):
    pass


b = B()

b.m1()
b.m2(5, 6)
```

```python
class A:
    def m1(self):
        print("hello")
    def m2(self):
        print("Hi")

class B(A):
    def m3(self):
        print("By")

class C(B):
    pass

c = C()

c.m1()
c.m2()
c.m3()
```

```python

class ShoppingCart:

    class Product:
        def __init__(self, name, price, qty):
            self.name = name
            self.price = price
            self.qty = qty

        def total(self):
            return self.price * self.qty

    def __init__(self):
        self.cart = []

    def add_product(self):
        name = input("Enter product name: ")
        price = int(input("Enter product price: "))
        qty = int(input("Enter no. of quantity: "))

        p = self.Product(name, price, qty)
        self.cart.append(p)

        print("Product Added Successfully")

    def show_cart(self):
        if len(self.cart) == 0:
            print("Cart is Empty")
        else:
            print("\n----- Shopping Cart -----")

            for c in self.cart:
                print("Name:", c.name)
                print("Price:", c.price)
                print("Quantity:", c.qty)
                print("Total:", c.total())
                print("-------------------------")

    def remove_product(self):
        name = input("Enter product name to remove: ")

        for c in self.cart:
            if c.name.lower() == name.lower():
                self.cart.remove(c)
                print("Product Removed Successfully")
                return

        print("Product Not Found")

    def grand_total(self):
        total = 0

        for c in self.cart:
            total += c.total()

        print("Grand Total:", total)


shop = ShoppingCart()

while True:

    print("\n===== SHOPPING CART =====")
    print("1. Add Product")
    print("2. Show Cart")
    print("3. Remove Product")
    print("4. Show Grand Total")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        shop.add_product()

    elif choice == 2:
        shop.show_cart()

    elif choice == 3:
        shop.remove_product()

    elif choice == 4:
        shop.grand_total()

    elif choice == 5:
        print("Thank you for using Shopping Cart")
        break

    else:
        print("Invalid Choice")

```
