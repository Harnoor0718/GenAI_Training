```python
class Test:
    # class level variable
    a = 10
    def __init__(self):
        # instance variable
        self.b = 20
t = Test()
print(t.a, t.b)

t.a = 11
t.b = 22
print(t.a, t.b)

t1 = Test()
print(t1.a, t1.b) # instance variable is varied from object to object
```

```python
class Demo:

    def msg(self):
        print("bye")

    @classmethod
    def show(cls):
        print("hello")

    @staticmethod
    def disp():
        print("hi")

d = Demo()

# We can access instance method by using object reference
d.msg()

# We can access classmethod by using object reference as well as class name
Demo.show()
d.show()

# we can access static method by using only class name
Demo.disp()



    
```

```python
# when used instance, classmethod, staticmethod
# self can get object information
# cls can get class information
# staticmethod can get only logic and calculation

# we can declare static/ class level variable inside class and constructor

class Test:
    count = 0

    def __init__(self, name):
        self.name = name
        Test.count += 1

t1 = Test("Ravi")
t2 = Test("Mukesh")
t3 = Test("Rahul")
print("Number of employees", Test.count)
```

```python
# diff bw instance vs static/ class vs local variable

class Student:

    # class level

    clg_name = "CGC"

    def __init__(self, name, rollNo):

        # instance level
        
        self.name = name
        self.rollNo = rollNo

    def show(self):
        greeting = "hello, " + self.name
        print(greeting)
        print(self.clg_name, self.name, self.rollNo)

    

s = Student("Harnoor", 89)
s.show()







        
```

```python
class A:

    def m1(self):
        print("hello")

    def m2(self, a, b):
        print(a + b)


class B(A):
    pass


b = B()

b.m1()
b.m2(5, 6)
```

```python
class A:
    def m1(self):
        print("hello")
    def m2(self):
        print("Hi")

class B(A):
    def m3(self):
        print("By")

class C(B):
    pass

c = C()

c.m1()
c.m2()
c.m3()
```

```python
class Parent:
    def marry(self):
        print("West Indies girl")

class Child(Parent):
    def marry(self):
        print("Indian girl")

c = Child()
c.marry()

```
