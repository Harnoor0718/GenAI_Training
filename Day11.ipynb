```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = {
    "study_hour" : [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "marks" : [23, 45, 57, 64, 69, 79, 88, 89, 97]
}

df = pd.DataFrame(data)
df
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.corr
```

```python
# data is divided into 2 parts for training experience
X = df[["study_hour"]]
y = df["marks"]
```

```python
# model build
model = LinearRegression()
model
```

```python
# model train
model.fit(X, y) # fit() is used for training experience
```

```python
c = model.intercept_
print(round(c, 2))
```

```python
# y = mx + c

m = model.coef_[0]
print(round(m, 2))
```

```python
# formula apply y = mx + c
study_hour = float(input("Enter study hours :"))
y = m * study_hour + c
print("Predicted Marks :", round(y, 2))

# when x = 0, y = c ; y = mx + c (y = 25.81 = c) 
```

```python
# Model predicted value

y_pred = model.predict(X)
y_pred
```

```python
compare_df = pd.DataFrame({
    "Actual x " : X.values.flatten(),
    "Actual y" : y.values,
    "Model Predicted Value" : y_pred
})

compare_df
```

```python
X.values
```

```python
y.values
```

```python
plt.plot(compare_df["Actual x "], compare_df["Model Predicted Value"], color = "r", marker = "o", label = "Model best line" )
plt.scatter(compare_df["Actual x "], compare_df["Actual y"], color = "g", label = "Original Data")
plt.title("Linear Regression Model")
plt.xlabel("study_hour----->")
plt.ylabel("Marks----->")
plt.legend()
plt.show()
```

```python
MAE = (compare_df["Actual y"] - compare_df["Model Predicted Value"]).abs().mean()
print("Mean Absolute Error", round(MAE, 2))
```

```python
MSE = ((compare_df["Actual y"] - compare_df["Model Predicted Value"]) ** 2).abs().mean()
print("Mean Squared Error", round(MSE, 2))
```

```python
import math
RMSE = math.sqrt(MSE)
print("Root Mean Squared Error :", RMSE)
```

```python
model_score = round(model.score(X, y), 2) * 100
print("After Training Model Score is :", model_score) 
```
