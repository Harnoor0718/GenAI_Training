```python
import numpy as np
import pandas as pd

emp = {
    "id" : [1, 2, 3, 4, 5],
    "name" : ["Harnoor", "Sahil", "Aditi", "Ram", "Shyam"],
    "age" : [12, 34, 56, 78, 22],
    "Dept" : ["CSE", "ECE", "CA", "BCA", "LAW"],
    "Salary" : [12000, 78000, 45000, 66000, 33000]
    }

df = pd.DataFrame(emp)
```

```python
print(df)
```

```python
print(df.loc[0])
```

```python
print(df.loc[2, "Dept"])
```

```python
# get particular column
print(df.loc[:,"name"])
```

```python
# get multiple columns

print(df.loc[:, ["name", "Salary"]])
```

```python
# get multiple rows

print(df.loc[[1, 2, 4]])
```

```python
df.loc[df["Salary"] > 30000]
```

```python
df.loc[df["Dept"] == "CA"]
```
