```python
from sklearn.preprocessing import StandardScaler
scaller = StandardScaler()
x_scaler = scaller.fit_transform(x)
x_train, x_test, y_train, y_test = train_test_split(x_scaler, y, test_size = 0.2, random_state = 43)
```

```python
from sklearn.metrics import r2_score
linear = LinearRegression()
linear.fit(x_train,y_train)
lin_pred = linear.predict(x_test)
lin_accuracy = r2_score(y_test, lin_pred) * 100
```

```python
ridge = Ridge(alpha = 0.1)
ridge.fit(x_train,y_train)
ridge_pred = ridge.predict(x_test)
ridge_accuracy = r2_score(y_test, ridge_pred)*100
```

```python
lasso = Lasso(alpha = 0.1)
lasso.fit(x_train,y_train)
lasso_pred = lasso.predict(x_test)
lasso_accuracy = r2_score(y_test, lasso_pred)*100
```

```python
print("Linear Regression Accuracy: ",round(lin_accuracy,2))
print("Ridge Regression Accuracy: ",round(ridge_accuracy,2))
print("Lasso Regression Accuracy: ",round(lasso_accuracy,2))
```

```python
final_compare = pd.DataFrame({
    "Actual" : y_test,
    "Ridge" : ridge_pred,
    "Lasso" : lasso_pred
})
final_compare
```

```python
# Final Conclusion
models = ["Linear", "Ridge" , "Lasso"]
score = [
    r2_score(y_test, lin_pred),
    r2_score(y_test, ridge_pred),
    r2_score(y_test, lasso_pred)
]
plt.bar(models,score)
plt.xlabel("Models")
plt.ylabel("Score")
plt.title("Linear vs Ridge vs Lasso")
plt.show()
```
