```python
from abc import ABC, abstractmethod
class Demo(ABC):
    @abstractmethod
    def show():
        pass

    @abstractmethod
    def msg():
        pass

class Test(Demo):
    def show(self):
        print("hello")

    def msg(self):
        print("hi")

t = Test()
t.show()
t.msg()
```

```python
from abc import ABC, abstractmethod

class Test1(ABC):
    @abstractmethod
    def m1():
        pass

class Test2(ABC):
    @abstractmethod
    def m2():
        pass

class Test(Test1, Test2):
    def m1(self):
        print("m1")

    def m2(self):
        print("m2")

t = Test()
t.m1()
t.m2()
    
```

```python
from abc import ABC, abstractmethod

class BankAccount:

    class InsufficientBalance(Exception):
        pass

    class BankAccount:
        def __init__(self, name, account_no, balance):
            self.name = name
            self.account_no = account_no
            self.balance = balance

    def deposit(self, amt):
        if amt < 0:
            print("Please deposit your amount")
        self.balance = self.balance + amt

        print("After Deposit Balance is:", self.balance)

    @abstractmethod
    def withdraw(self, amt):
        pass

    def show_details(self):
        print("Customer Name :", self.name)
        print("Account No :", self.account_no)
        print("Your Balance is ", self.balance)


class Customer(BankAccount):

    def withdraw(self, amt):

        if amt > self.balance:
            raise BankAccount.InsufficientBalance("Insufficient Balance")

        amt = int(input("Enter amount to withdraw: "))
        self.balance = self.balance - amt
        print("Withdrawal Successful")
        print("New Balance is :", self.balance)


name = input("Enter Customer Name: ")
account_no = input("Enter Account Number: ")
balance = int(input("Enter Initial Balance: "))

c = Customer()
c.name = name
c.account_no = account_no
c.balance = balance

while True:
    print("\n===== Bank Account =====")
    print("1. View Details")
    print("2. View Balance")
    print("3. Withdraw")
    print("4. Deposit")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        c.show_details()

    elif choice == 2:
        print("Your Balance is:", c.balance)

    elif choice == 3:
        try:
            amt = int(input("Enter amount to withdraw: "))

            if amt > c.balance:
                raise BankAccount.InsufficientBalance("Insufficient Balance")

            c.balance = c.balance - amt
            print("Withdrawal Successful")
            print("New Balance is:", c.balance)

        except BankAccount.InsufficientBalance as e:
            print(e)

    elif choice == 4:
        amt = int(input("Enter amount to deposit: "))
        c.deposit(amt)

    elif choice == 5:
        print("Thank you for using Bank Account System")
        break

    else:
        print("Invalid Choice")
```
