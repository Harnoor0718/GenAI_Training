```python
df = pd.DataFrame({
    "Names" : ["Ravi", "Suresh", "Sita", "Sonu", "Gita", "Mukesh", "Rita"],
    "Gender" : ["Male", "Male", "Female", "Male", "Female", "Male", "Female"]
})
df
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    "City" : ["Delhi", "Mumbai", "Noida", "Delhi", "Mumbai", "Gurgaon"]
})
df
```

```python
encoder = OneHotEncoder()
encoder
```

```python
encoder_array = encoder.fit_transform(df[["City"]])
encoder_array
```

```python
cols_name = encoder.get_feature_names_out(["City"])
cols_name
```
