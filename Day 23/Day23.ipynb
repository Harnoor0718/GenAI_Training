```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    "City" : ["Delhi", "Mumbai", "Noida", "Delhi", "Mumbai", "Gurgaon"]
})
df
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
encoder
```

```python
encoder_array = encoder.fit_transform(df[["City"]])
encoder_array
```

```python
cols_name = encoder.get_feature_names_out(["City"])
cols_name
```

```python
result = pd.DataFrame(encoder_array, columns=cols_name, dtype=int)
result
```

```python
df = pd.DataFrame({
    "Name" : ["Harnoor", "Sahil", "Tanveer", "Ram", "Sham", "Sita", "Gita"],
    "Dept" : ["CSE", "ECE", "BCA", "BBA", "BBA", "CSE", "ECE"],
    "City" : ["Amritsar", "Ludhiana", "Mohali", "Chandiagrh", "Ludhiana", "Mohali", "Mohali"],
    "Age" : [23, 21, 34, 21, 23, 33, 23]
})
df
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
encoder
```

```python
encoder_array = encoder.fit_transform(df[["Dept", "City"]])
encoder_array
```

```python
cols_name = encoder.get_feature_names_out(["Dept", "City"])
cols_name
```

```python
result = pd.DataFrame(encoder_array, columns=cols_name, dtype=int)
result
```

```python
result = pd.DataFrame(encoder_array, columns=cols_name, dtype=int)
result = pd.concat([result, df[["Age"]].reset_index(drop=True)], axis=1)
result
```
